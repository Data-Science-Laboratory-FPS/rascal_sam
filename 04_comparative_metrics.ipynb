## Step 4. Comparative Metrics and Statistical Analysis

This notebook aggregates and compares results from the three segmentation experiments.
It requires the `experiments/` folder generated by `03_run_experiments.ipynb`.

### Metrics reported
- **Sensitivity (Recall)** and **Specificity**: primary clinical metrics
- **Balanced Accuracy**: main summary metric given the 10:1 class imbalance
- **AUC-ROC** and **AUPRC**: discrimination ability
- **F2-score**: weights recall twice as much as precision, appropriate for
  mortality detection where missing a death is more costly than a false alarm
- **F1-score**, **Precision**, **MCC**: reported for completeness

All metrics reported as mean ± std across folds and with **bootstrap 95% CI**
(n = 1,000 resamples) on concatenated out-of-fold predictions.

### Statistical comparison of AUC
- **DeLong test**: standard parametric test for paired AUC comparison
- **Paired bootstrap**: non-parametric comparison, more robust with small
  sample sizes (n = 98 positive events)

### Visualisations
- Boxplots across folds (6 key metrics)
- ROC curves comparison
- Precision-Recall curves comparison
- Aggregated confusion matrices per segmentation type

> **Outputs:** `metrics_summary.csv`, `metrics_summary.tex`,
> `auc_comparison_delong.csv`, `auc_comparison_bootstrap_paired.csv`,
> `fig_boxplots_metrics.pdf/png`, `fig_roc_curves.pdf/png`,
> `fig_pr_curves.pdf/png`, `fig_confusion_matrices.pdf/png`

```python
import os
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

from sklearn.metrics import (
    precision_score, recall_score,
    precision_recall_curve, auc,
    matthews_corrcoef, fbeta_score,
    roc_curve, average_precision_score,
    roc_auc_score, confusion_matrix
)
from itertools import combinations
from scipy import stats

# ===============================
# GLOBAL FIGURE CONFIGURATION
# ===============================
plt.rcParams.update({
    "font.family":    "sans-serif",
    "font.size":      11,
    "axes.titlesize": 12,
    "axes.labelsize": 11,
    "legend.fontsize": 10,
    "figure.dpi":     150,
})

BASE_DIR = "experiments"

MODEL_LABELS = {
    "experts":  "Expert segmentation",
    "sam_bbox": "SAM (bounding box)",
    "sam_mask": "SAM (mask)"
}

MODEL_ORDER = ["experts", "sam_bbox", "sam_mask"]

# Colour-blind safe palette
PALETTE = {
    "experts":  "#0077BB",
    "sam_bbox": "#EE7733",
    "sam_mask": "#009988"
}


# ===============================
# BOOTSTRAP 95% CI
# ===============================
def bootstrap_ci(y_true, y_probs, threshold, n_bootstrap=1000, ci=0.95, seed=42):
    """
    Computes bootstrap 95% CI for all classification metrics.
    Returns a dict with mean, ci_low, and ci_high per metric.
    """
    rng = np.random.RandomState(seed)
    n   = len(y_true)
    metrics_boot = {
        "Recall": [], "Specificity": [], "BalAcc": [],
        "Precision": [], "F1": [], "F2": [],
        "MCC": [], "AUC": [], "AUPRC": []
    }

    for _ in range(n_bootstrap):
        idx = rng.randint(0, n, n)
        yt  = y_true[idx]
        yp  = y_probs[idx]

        if len(np.unique(yt)) < 2:
            continue

        yp_bin = (yp >= threshold).astype(int)
        sens   = recall_score(yt, yp_bin, zero_division=0)
        spec   = recall_score(1 - yt, 1 - yp_bin, zero_division=0)

        metrics_boot["Recall"].append(sens)
        metrics_boot["Specificity"].append(spec)
        metrics_boot["BalAcc"].append((sens + spec) / 2)
        metrics_boot["Precision"].append(precision_score(yt, yp_bin, zero_division=0))
        metrics_boot["F1"].append(fbeta_score(yt, yp_bin, beta=1, zero_division=0))
        metrics_boot["F2"].append(fbeta_score(yt, yp_bin, beta=2, zero_division=0))
        metrics_boot["MCC"].append(matthews_corrcoef(yt, yp_bin))
        metrics_boot["AUC"].append(roc_auc_score(yt, yp))
        metrics_boot["AUPRC"].append(average_precision_score(yt, yp))

    alpha   = (1 - ci) / 2
    results = {}
    for metric, values in metrics_boot.items():
        values = np.array(values)
        results[metric] = {
            "mean":    np.mean(values),
            "ci_low":  np.percentile(values, alpha * 100),
            "ci_high": np.percentile(values, (1 - alpha) * 100)
        }
    return results


# ===============================
# LOAD METRICS PER FOLD
# ===============================
rows = []

for model_name in MODEL_ORDER:
    model_path = os.path.join(BASE_DIR, model_name)

    if not os.path.isdir(model_path):
        print(f"[WARNING] Not found: {model_path}")
        continue

    for fold_folder in sorted(os.listdir(model_path)):
        fold_path = os.path.join(model_path, fold_folder)

        if not os.path.isdir(fold_path):
            continue

        y_true_path    = os.path.join(fold_path, "y_true.npy")
        y_probs_path   = os.path.join(fold_path, "y_probs.npy")
        threshold_path = os.path.join(fold_path, "threshold.npy")

        if not (os.path.exists(y_true_path) and os.path.exists(y_probs_path)):
            continue

        y_true    = np.load(y_true_path)
        y_probs   = np.load(y_probs_path)
        threshold = float(np.load(threshold_path)) if os.path.exists(threshold_path) else 0.5

        y_pred = (y_probs >= threshold).astype(int)
        sens   = recall_score(y_true, y_pred, zero_division=0)
        spec   = recall_score(1 - y_true, 1 - y_pred, zero_division=0)

        rows.append({
            "Model":       model_name,
            "ModelLabel":  MODEL_LABELS.get(model_name, model_name),
            "Fold":        fold_folder,
            "Threshold":   threshold,
            "Recall":      sens,
            "Specificity": spec,
            "BalAcc":      (sens + spec) / 2,
            "Precision":   precision_score(y_true, y_pred, zero_division=0),
            "F1":          fbeta_score(y_true, y_pred, beta=1, zero_division=0),
            "F2":          fbeta_score(y_true, y_pred, beta=2, zero_division=0),
            "MCC":         matthews_corrcoef(y_true, y_pred),
            "AUC":         roc_auc_score(y_true, y_probs),
            "AUPRC":       average_precision_score(y_true, y_probs),
        })

df_metrics = pd.DataFrame(rows)

print("\nMETRICS PER FOLD:\n")
display(df_metrics.drop(columns=["Model"]))


# ===============================
# GLOBAL SUMMARY MEAN ± STD
# ===============================
metric_cols = ["Recall", "Specificity", "BalAcc",
               "Precision", "F1", "F2", "MCC", "AUC", "AUPRC"]

summary = df_metrics.groupby("ModelLabel")[metric_cols].agg(["mean", "std"])
print("\nGLOBAL SUMMARY (MEAN ± STD):\n")
display(summary)

summary.to_csv("metrics_summary.csv")
summary.to_latex("metrics_summary.tex")


# ===============================
# BOOTSTRAP 95% CI PER MODEL
# ===============================
print("\nBOOTSTRAP 95% CI PER MODEL:")
print(f"\n{'Model':<25} {'Metric':<14} {'Mean':>8} {'95% CI':>20}")
print("-" * 72)

ci_all = {}
for model_name in MODEL_ORDER:
    model_path = os.path.join(BASE_DIR, model_name)
    if not os.path.isdir(model_path):
        continue

    all_y, all_p, all_t = [], [], []
    for fold_folder in sorted(os.listdir(model_path)):
        fold_path = os.path.join(model_path, fold_folder)
        if not os.path.isdir(fold_path):
            continue
        yf = os.path.join(fold_path, "y_true.npy")
        pf = os.path.join(fold_path, "y_probs.npy")
        tf = os.path.join(fold_path, "threshold.npy")
        if os.path.exists(yf) and os.path.exists(pf):
            all_y.extend(np.load(yf))
            all_p.extend(np.load(pf))
            if os.path.exists(tf):
                all_t.append(float(np.load(tf)))

    all_y            = np.array(all_y)
    all_p            = np.array(all_p)
    global_threshold = np.mean(all_t) if all_t else 0.5

    ci = bootstrap_ci(all_y, all_p, threshold=global_threshold)
    ci_all[model_name] = ci

    label = MODEL_LABELS.get(model_name, model_name)
    for metric, vals in ci.items():
        print(f"{label:<25} {metric:<14} {vals['mean']:>8.3f} "
              f"[{vals['ci_low']:.3f}, {vals['ci_high']:.3f}]")
    print()


# ===============================
# FIGURE 1: BOXPLOTS — KEY METRICS
# ===============================
metrics_to_plot = ["Recall", "Specificity", "BalAcc", "AUC", "F2", "MCC"]
metric_titles   = ["Sensitivity (Recall)", "Specificity",
                   "Balanced Accuracy", "AUC-ROC", "F2-score", "MCC"]

fig, axes = plt.subplots(2, 3, figsize=(13, 8))
axes = axes.flatten()

label_order = [MODEL_LABELS[m] for m in MODEL_ORDER if m in df_metrics["Model"].unique()]

for i, (metric, title) in enumerate(zip(metrics_to_plot, metric_titles)):
    ax = axes[i]
    sns.boxplot(
        data=df_metrics,
        x="ModelLabel", y=metric,
        order=label_order,
        palette=[PALETTE[m] for m in MODEL_ORDER if m in df_metrics["Model"].unique()],
        width=0.5, linewidth=1.2,
        ax=ax
    )
    ax.set_title(title, fontweight="bold")
    ax.set_xlabel("")
    ax.set_ylabel(title)
    ax.set_ylim(-0.05, 1.05)
    ax.tick_params(axis="x", rotation=15)
    ax.yaxis.set_major_formatter(mticker.FormatStrFormatter("%.2f"))
    ax.grid(axis="y", alpha=0.3, linestyle="--")
    sns.despine(ax=ax)

plt.suptitle("Model comparison across segmentation types (4-fold CV)",
             fontsize=13, fontweight="bold", y=1.01)
plt.tight_layout()
plt.savefig("fig_boxplots_metrics.pdf", bbox_inches="tight")
plt.savefig("fig_boxplots_metrics.png", bbox_inches="tight", dpi=300)
plt.show()


# ===============================
# FIGURE 2: ROC CURVES
# ===============================
fig, ax = plt.subplots(figsize=(6, 5.5))

for model_name in MODEL_ORDER:
    model_path = os.path.join(BASE_DIR, model_name)
    if not os.path.isdir(model_path):
        continue

    all_y, all_p = [], []
    for fold_folder in sorted(os.listdir(model_path)):
        fold_path = os.path.join(model_path, fold_folder)
        if not os.path.isdir(fold_path):
            continue
        yf = os.path.join(fold_path, "y_true.npy")
        pf = os.path.join(fold_path, "y_probs.npy")
        if os.path.exists(yf) and os.path.exists(pf):
            all_y.extend(np.load(yf))
            all_p.extend(np.load(pf))

    all_y    = np.array(all_y)
    all_p    = np.array(all_p)
    fpr, tpr, _ = roc_curve(all_y, all_p)
    roc_auc  = auc(fpr, tpr)
    label    = MODEL_LABELS.get(model_name, model_name)

    ax.plot(fpr, tpr,
            color=PALETTE[model_name],
            linewidth=2,
            label=f"{label} (AUC = {roc_auc:.3f})")

ax.plot([0, 1], [0, 1], linestyle="--", color="gray",
        linewidth=1, label="Random classifier")
ax.set_xlabel("False Positive Rate (1 – Specificity)")
ax.set_ylabel("True Positive Rate (Sensitivity)")
ax.set_title("ROC curves — segmentation comparison", fontweight="bold")
ax.legend(loc="lower right")
ax.grid(alpha=0.3, linestyle="--")
sns.despine(ax=ax)
plt.tight_layout()
plt.savefig("fig_roc_curves.pdf", bbox_inches="tight")
plt.savefig("fig_roc_curves.png", bbox_inches="tight", dpi=300)
plt.show()


# ===============================
# FIGURE 3: PRECISION–RECALL CURVES
# More informative than ROC under class imbalance (10:1)
# ===============================
fig, ax = plt.subplots(figsize=(6, 5.5))

for model_name in MODEL_ORDER:
    model_path = os.path.join(BASE_DIR, model_name)
    if not os.path.isdir(model_path):
        continue

    all_y, all_p = [], []
    for fold_folder in sorted(os.listdir(model_path)):
        fold_path = os.path.join(model_path, fold_folder)
        if not os.path.isdir(fold_path):
            continue
        yf = os.path.join(fold_path, "y_true.npy")
        pf = os.path.join(fold_path, "y_probs.npy")
        if os.path.exists(yf) and os.path.exists(pf):
            all_y.extend(np.load(yf))
            all_p.extend(np.load(pf))

    all_y = np.array(all_y)
    all_p = np.array(all_p)

    precision, recall, _ = precision_recall_curve(all_y, all_p)
    auprc = average_precision_score(all_y, all_p)
    label = MODEL_LABELS.get(model_name, model_name)

    ax.plot(recall, precision,
            color=PALETTE[model_name],
            linewidth=2,
            label=f"{label} (AUPRC = {auprc:.3f})")

# Random baseline = positive class prevalence
pos_rate = sum(np.load(os.path.join(BASE_DIR, "experts", "fold_0", "y_true.npy"))) / \
           len(np.load(os.path.join(BASE_DIR, "experts", "fold_0", "y_true.npy"))) \
           if os.path.exists(os.path.join(BASE_DIR, "experts", "fold_0", "y_true.npy")) else 0.09
ax.axhline(y=pos_rate, linestyle="--", color="gray",
           linewidth=1, label=f"Random classifier (prevalence = {pos_rate:.2f})")

ax.set_xlabel("Recall (Sensitivity)")
ax.set_ylabel("Precision")
ax.set_title("Precision–Recall curves — segmentation comparison", fontweight="bold")
ax.legend(loc="upper right")
ax.grid(alpha=0.3, linestyle="--")
sns.despine(ax=ax)
plt.tight_layout()
plt.savefig("fig_pr_curves.pdf", bbox_inches="tight")
plt.savefig("fig_pr_curves.png", bbox_inches="tight", dpi=300)
plt.show()


# ===============================
# FIGURE 4: AGGREGATED CONFUSION MATRICES
# ===============================
fig, axes = plt.subplots(1, 3, figsize=(13, 4))

for i, model_name in enumerate(MODEL_ORDER):
    model_path = os.path.join(BASE_DIR, model_name)
    if not os.path.isdir(model_path):
        continue

    all_y, all_pred = [], []
    for fold_folder in sorted(os.listdir(model_path)):
        fold_path = os.path.join(model_path, fold_folder)
        if not os.path.isdir(fold_path):
            continue
        yf = os.path.join(fold_path, "y_true.npy")
        pf = os.path.join(fold_path, "y_probs.npy")
        tf = os.path.join(fold_path, "threshold.npy")
        if os.path.exists(yf) and os.path.exists(pf):
            yt  = np.load(yf)
            yp  = np.load(pf)
            thr = float(np.load(tf)) if os.path.exists(tf) else 0.5
            all_y.extend(yt)
            all_pred.extend((yp >= thr).astype(int))

    cm    = confusion_matrix(all_y, all_pred)
    label = MODEL_LABELS.get(model_name, model_name)

    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                xticklabels=["Alive", "Deceased"],
                yticklabels=["Alive", "Deceased"],
                ax=axes[i], cbar=False, linewidths=0.5)
    axes[i].set_title(label, fontweight="bold")
    axes[i].set_xlabel("Predicted")
    axes[i].set_ylabel("True" if i == 0 else "")

plt.suptitle("Aggregated confusion matrices (4-fold CV)",
             fontsize=13, fontweight="bold")
plt.tight_layout()
plt.savefig("fig_confusion_matrices.pdf", bbox_inches="tight")
plt.savefig("fig_confusion_matrices.png", bbox_inches="tight", dpi=300)
plt.show()


# ===============================
# DELONG TEST AND PAIRED BOOTSTRAP
#
# Two complementary methods for comparing AUC across models:
#
# 1. DeLong test: standard parametric test for paired AUC comparison.
#    Returns z-score and p-value.
#
# 2. Paired bootstrap: non-parametric comparison resampling the same
#    indices for both models. More robust with small sample sizes.
# ===============================

def delong_auc_variance(y_true, y_probs):
    """
    Computes AUC variance using the DeLong (1988) method.
    Required for the DeLong test between two AUCs.
    """
    n_pos = int(y_true.sum())
    n_neg = int((1 - y_true).sum())

    pos_probs = y_probs[y_true == 1]
    neg_probs = y_probs[y_true == 0]

    def placement_values(a, b):
        """P(score_a > score_b) + 0.5 * P(score_a == score_b)"""
        result = np.zeros(len(a))
        for i, ai in enumerate(a):
            result[i] = np.mean(ai > b) + 0.5 * np.mean(ai == b)
        return result

    V_pos = placement_values(pos_probs, neg_probs)
    V_neg = 1 - placement_values(neg_probs, pos_probs)

    var = (np.var(V_pos) / n_pos) + (np.var(V_neg) / n_neg)
    return var


def delong_test(y_true, y_probs_a, y_probs_b):
    """
    DeLong test for comparing two AUCs on the same observations.
    Returns: auc_a, auc_b, difference, z-score, two-tailed p-value.
    """
    auc_a = roc_auc_score(y_true, y_probs_a)
    auc_b = roc_auc_score(y_true, y_probs_b)

    var_a = delong_auc_variance(y_true, y_probs_a)
    var_b = delong_auc_variance(y_true, y_probs_b)

    # Simplified SE assuming independence (standard approximation for small n)
    se = np.sqrt(var_a + var_b)

    if se == 0:
        return auc_a, auc_b, auc_a - auc_b, 0.0, 1.0

    z = (auc_a - auc_b) / se
    p = 2 * (1 - stats.norm.cdf(abs(z)))

    return auc_a, auc_b, auc_a - auc_b, z, p


def bootstrap_paired_auc(y_true, y_probs_a, y_probs_b,
                          n_bootstrap=1000, seed=42):
    """
    Paired bootstrap for AUC comparison.
    Resamples the same indices for both models (paired).
    Returns: mean difference, 95% CI, empirical p-value.
    """
    rng   = np.random.RandomState(seed)
    n     = len(y_true)
    diffs = []

    for _ in range(n_bootstrap):
        idx = rng.randint(0, n, n)
        yt  = y_true[idx]
        ya  = y_probs_a[idx]
        yb  = y_probs_b[idx]

        if len(np.unique(yt)) < 2:
            continue

        diffs.append(roc_auc_score(yt, ya) - roc_auc_score(yt, yb))

    diffs     = np.array(diffs)
    diff_mean = np.mean(diffs)
    ci_low    = np.percentile(diffs, 2.5)
    ci_high   = np.percentile(diffs, 97.5)

    # Empirical p-value: proportion of times the difference crosses zero
    p_emp = min(np.mean(diffs <= 0), np.mean(diffs >= 0)) * 2

    return diff_mean, ci_low, ci_high, p_emp


# ===============================
# LOAD CONCATENATED PROBABILITIES PER MODEL
# ===============================
probs_by_model = {}
ytrue_by_model = {}

for model_name in MODEL_ORDER:
    model_path = os.path.join(BASE_DIR, model_name)
    if not os.path.isdir(model_path):
        continue

    all_y, all_p = [], []
    for fold_folder in sorted(os.listdir(model_path)):
        fold_path = os.path.join(model_path, fold_folder)
        if not os.path.isdir(fold_path):
            continue
        yf = os.path.join(fold_path, "y_true.npy")
        pf = os.path.join(fold_path, "y_probs.npy")
        if os.path.exists(yf) and os.path.exists(pf):
            all_y.extend(np.load(yf))
            all_p.extend(np.load(pf))

    probs_by_model[model_name] = np.array(all_p)
    ytrue_by_model[model_name] = np.array(all_y)


# ===============================
# PAIRWISE AUC COMPARISON TABLE
# ===============================
pairs = list(combinations(MODEL_ORDER, 2))

print("\n" + "="*80)
print("AUC COMPARISON ACROSS MODELS")
print("="*80)

print("\n--- DeLong TEST ---")
print(f"{'Pair':<45} {'AUC_A':>7} {'AUC_B':>7} {'Diff':>7} {'Z':>7} {'p-value':>10} {'Sig':>5}")
print("-" * 95)

delong_results = []
for model_a, model_b in pairs:
    if model_a not in probs_by_model or model_b not in probs_by_model:
        continue

    y_true = ytrue_by_model[model_a]
    ya     = probs_by_model[model_a]
    yb     = probs_by_model[model_b]

    auc_a, auc_b, diff, z, p = delong_test(y_true, ya, yb)
    sig      = "**" if p < 0.01 else ("*" if p < 0.05 else "ns")
    pair_str = f"{MODEL_LABELS[model_a]} vs {MODEL_LABELS[model_b]}"

    print(f"{pair_str:<45} {auc_a:>7.3f} {auc_b:>7.3f} {diff:>+7.3f} {z:>7.2f} {p:>10.4f} {sig:>5}")
    delong_results.append({
        "Model_A": model_a, "Model_B": model_b,
        "AUC_A": auc_a, "AUC_B": auc_b,
        "Diff": diff, "Z": z, "p_delong": p, "Sig_delong": sig
    })

print("\n* p<0.05  ** p<0.01  ns = not significant")

print("\n--- PAIRED BOOTSTRAP (n=1,000) ---")
print(f"{'Pair':<45} {'Diff':>7} {'CI95% low':>12} {'CI95% high':>12} {'p-emp':>10} {'Sig':>5}")
print("-" * 95)

boot_results = []
for model_a, model_b in pairs:
    if model_a not in probs_by_model or model_b not in probs_by_model:
        continue

    y_true = ytrue_by_model[model_a]
    ya     = probs_by_model[model_a]
    yb     = probs_by_model[model_b]

    diff_mean, ci_low, ci_high, p_emp = bootstrap_paired_auc(y_true, ya, yb)
    sig      = "**" if p_emp < 0.01 else ("*" if p_emp < 0.05 else "ns")
    pair_str = f"{MODEL_LABELS[model_a]} vs {MODEL_LABELS[model_b]}"

    print(f"{pair_str:<45} {diff_mean:>+7.3f} {ci_low:>12.3f} {ci_high:>12.3f} {p_emp:>10.4f} {sig:>5}")
    boot_results.append({
        "Model_A": model_a, "Model_B": model_b,
        "Diff_mean": diff_mean, "CI_low": ci_low,
        "CI_high": ci_high, "p_bootstrap": p_emp, "Sig_bootstrap": sig
    })

print("\n* p<0.05  ** p<0.01  ns = not significant")

# Save comparison results
df_delong = pd.DataFrame(delong_results)
df_boot   = pd.DataFrame(boot_results)

df_delong.to_csv("auc_comparison_delong.csv", index=False)
df_boot.to_csv("auc_comparison_bootstrap_paired.csv", index=False)

print("\nResults saved:")
print("  auc_comparison_delong.csv")
print("  auc_comparison_bootstrap_paired.csv")
```
