## Step 3. Run experiments

This notebook runs the classification pipeline for each of the three
segmentation conditions. It requires:
- `label_clasification/df_all.csv` generated by `01_data_organisation.ipynb`
- The functions defined in `02_core_pipeline.ipynb`

The three experiments use **identical pipeline configuration** and the same
4-fold splits (seed 42), ensuring results are directly comparable.


```python
from core_pipeline import run_pipeline_4fold

df_all = pd.read_csv("label_clasification/df_all.csv")
```

## Step 3a — Expert Segmentation
Pipeline run on images segmented by expert radiologists (manual bounding boxes).
Results saved to `experiments/experts/`.

```python
import os
import pandas as pd

# ===============================
# Filter dataframe for experts
# ===============================
df_exp = df_all[df_all["segmentation"] == "experts"].copy()

base_dir = "label_clasification"

# ===============================
# RUN 4-FOLD PIPELINE
# ===============================
df_results, ci_results = run_pipeline_4fold(
    df_exp,
    base_dir,
    experiment_name="experts",
    batch_size=8
)

# ===============================
# CONFUSION MATRIX por fold
# ===============================
print("\nConfusion matrices per fold:")
for _, row in df_results.iterrows():
    print(f"\nFold {int(row['fold'])}:")
    print(row["ConfMatrix"])
```
